```python
def percentage(a,b,c,d,e):
    print("Subject1=",a);
    print("Subject2=",b);
    print("Subject3=",c);
    print("Subject4=",d);
    print("Subject5=",e);
    total=a+b+c+d+e;
    average=total/5;
    print("Total :",total);
    print("Percentage : ",average);
 
```

```python
percentage(98,87,95,95,93)
```
